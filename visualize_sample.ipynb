```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sample_input = pd.read_csv("sampledata/sampledata.csv", header=None).to_numpy()
```

```python
# delete the first row - outlier
sample_input = sample_input[1:, :]
sample_input
```

```python
num_of_channels = len(sample_input)
channel_len_input = len(sample_input[0])
```

```python
sample_output = pd.read_csv("sampledata/outputsample.csv", header=None).to_numpy()
sample_output = sample_output[1:, :]
```

```python
channel_len_output = len(sample_output[0])
```

```python
fig, axes = plt.subplots(num_of_channels, 2, figsize=(10, 2 * num_of_channels))

x_in = np.arange(0, 1, 1.0 / channel_len_input)
x_out = np.arange(0, 1, 1.0 / channel_len_output)

for i in range(num_of_channels):
    ax_in = axes[i, 0] if num_of_channels > 1 else axes[0]
    ax_out = axes[i, 1] if num_of_channels > 1 else axes[1]

    mean_in = np.mean(sample_input[i])
    mean_out = np.mean(sample_output[i])

    ax_in.plot(x_in, sample_input[i])
    ax_in.set_title(f'Channel {i+1}, mean: {mean_in:.2f}')
    
    ax_out.plot(x_out, sample_output[i])
    ax_out.set_title(f'Channel {i+1}, mean: {mean_out:.2f}')

fig.text(0.25, 0.999, 'Before', ha='center', va='bottom', fontsize=14)
fig.text(0.75, 0.999, 'After', ha='center', va='bottom', fontsize=14)

fig.suptitle('Visualization of sample data, EEG graph before and after ART', fontsize=16, y=1.005)

plt.tight_layout()
plt.show()
```
